### **List (Creation, Modifi cation and Access):**

**1. List Creation:**

```python
# a. Create a list named age_list with five integer elements.
age_list = [24, 25, 26, 27, 28]
print(age_list)
```

```python
# b. Create a list named name_list with five string elements.
name_list = ["Riddhi","Poornima","Anish","Chandran","Pooja"]
print(name_list)
```

**2. List Operations / Modifications:**

```python
# a. Append the string "Yazhini" to name_list.

name_list.append("Yazhini")
print(name_list)

# adds new element at the last position of the list.
```

```python
# b. Insert the element 30 at index 2 in age_list.

insert_age =30
age_list.insert(2,insert_age)
print(age_list)
# adds item at specific postion.
```

```python
# c. Remove the string "Yazhini" from name_list.

name_list.remove("Yazhini")
print(name_list)

# remove by value.
```

```python
# d. Pop the last element from age_list.

age_list.pop()
print(age_list)

# without index removes last element.
```

```python
# e. Extend the age_list with additional ages [29, 30, 26].

age_list.extend([29,30,26])
print(age_list)

# add multiple item.
```

```python
# f. Sort age_list in descending order.

age_list.sort(reverse=True)
print(age_list)

# reverse give descending order.
```

```python
# g. Find Max age, Min age and sum of all ages from age_list.

max_age     = max(age_list) # shows maximum value in the list.
min_age     = min(age_list) # shows minimum value in the list.
sum_of_ages = sum(age_list) # sum of ages.
print("Max_age :",max_age)
print("Min_age :",min_age)
print("Sum_of_ages :",sum_of_ages)
```

**3. Accessing List Elements:**

```python
# a. Print the first element of name_list.
name_list[0]
print(name_list[0])

```

```python
# b. Print the last element of name_list.
name_list[-1]
print(name_list[-1])
```

```python
# c. Print the elements from index 2 to index 4 in name_list.
name_list[2:5]
print(name_list[2:5])
```

```python
# d. Print the elements of name_list in reverse order.
name_list[::-1]
print(name_list[::-1])
```

### **Dictionary (Creation, Modification and Access):**

```python
# a. Create a dictionary named student_marks that maps the names of five students to their marks (use scale of from 0 to 100).
Student_marks ={
    "Riddhi": 88,
    "Poornima": 75,
    "Anish": 70,
    "Chandran" : 80,
    "Pooja" : 65
    }
print(Student_marks)
print(type(Student_marks))
```

```python
# b. Access and print the mark of a specific student, of your choice.
print(Student_marks["Riddhi"])
```

```python
# c. Add a new student "Janani" with a mark of 80 to the student_marks dictionary.
Student_marks["Janani"] = 80
print(Student_marks)
```

```python
# d. Update the mark of any one older student to 82.
Student_marks["Anish"] = 82
print(Student_marks)
```

```python
# e. Use the keys(), values(), and items() methods to print all keys, values, and key-value pairs in the student_marks dictionary.

print(Student_marks.keys()) # returns all keys.

print(Student_marks.values()) # returns all values.

print(Student_marks.items()) # returns keys-value pairs as tuple.
```

### **Sets (Operations):**

```python
# a. Create a set called my_set with following values: ['a','e','i','o','u','a','a','i']
my_set = {'a','e','i','o','u','a','a','i'}
print(my_set)
# They are unique, unordered items. Duplicate values are automatically removed.
```

```python
# b. Attempt to change the value of my_set[4] = 's'.
my_set =  {'a','e','i','o','u','a','a','i'}
my_set[4] = 's'
print(my_set)
# Shows error : TypeError
# Sets do not support indexing (unordered)
# cannot be accessed or modified using postions.
```

```python
# c. Create two sets:
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print(set1)
print(set2)
```

```python
# d. Compute and print the union and intersection of set1 and set2

union_set = set1.union(set2)
print(union_set)
# Union returns all unique elements from both sets.

intersection_set = set1.intersection(set2)
print(intersection_set)
# Intersection returns only elements present in both sets.
```

**Performance Category Program:**

```python
# 1. Prompt user for Input. Score range should be from 0 to 10 (both inclusive).
score = int(input("Enter your score (0 to 10): "))

# 2. Find the performance category based on the input score using following criteria:
# a. Above Average: Score greater than 7

if score < 0 or score > 10:
 print("Invalid input! Score should be between 0 and 10.")
elif score > 7:
   print("Above Average: Excellent work! Keep it up.")
# b. Average: Score between 4 and 7(both inclusive)
elif 4 <= score <= 7:
   print("Average: Good effort! Keep practicing, there's room for improvement.")
# c. Below Average: Score lesser than 4
else:
    print("Below Average: Need to improve your performance. Consistent practice will lead to better results.")
# 3. Output: Print the Performance category
```
